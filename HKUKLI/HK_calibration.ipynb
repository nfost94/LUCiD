```python
import numpy as np

r = np.load('calibration_results.npz')
FIELDS = list(r['fields'])
NS, R, H = int(r['ns']), float(r['r']), float(r['h'])
sigma, truth, start, fit_theta = r['sigma'], r['truth'], r['start'], r['fit_theta']

print(f'{NS} PMTs | r={R:.1f} m | H={H:.1f} m')

print('CRB (fractional 1σ) per parameter:')
for f, s in zip(FIELDS, sigma):
    print(f'  {f:22s} {s:6.2%}')

print(f'{"param":22s}{"truth":>9s}{"start":>9s}{"fit":>9s}{"err":>8s}{"CRB":>7s}')
for i, f in enumerate(FIELDS):
    e = fit_theta[i] / truth[i] - 1
    print(f'{f:22s}{truth[i]:9.3f}{start[i]:9.3f}{fit_theta[i]:9.3f}{e:+7.1%}{sigma[i]:7.1%}')
```
